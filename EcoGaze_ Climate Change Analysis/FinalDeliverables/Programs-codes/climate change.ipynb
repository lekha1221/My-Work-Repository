```python
import pandas as pd
```

```python
csv_file_path = r"C:\Users\Lenovo\Downloads\archive (1)\climate_change.csv"
climate_data = pd.read_csv(csv_file_path)
climate_data.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for seaborn
sns.set(style="whitegrid")

# Plot trends over time for greenhouse gases and temperature anomalies
fig, axs = plt.subplots(3, 2, figsize=(15, 18))

# CO2 concentration over time
sns.lineplot(x='Year', y='CO2', data=climate_data, ax=axs[0, 0])
axs[0, 0].set_title('CO2 Concentration Over Time')

# CH4 concentration over time
sns.lineplot(x='Year', y='CH4', data=climate_data, ax=axs[0, 1])
axs[0, 1].set_title('CH4 Concentration Over Time')

# N2O concentration over time
sns.lineplot(x='Year', y='N2O', data=climate_data, ax=axs[1, 0])
axs[1, 0].set_title('N2O Concentration Over Time')

# CFC-11 concentration over time
sns.lineplot(x='Year', y='CFC-11', data=climate_data, ax=axs[1, 1])
axs[1, 1].set_title('CFC-11 Concentration Over Time')

# CFC-12 concentration over time
sns.lineplot(x='Year', y='CFC-12', data=climate_data, ax=axs[2, 0])
axs[2, 0].set_title('CFC-12 Concentration Over Time')

# Temperature anomaly over time
sns.lineplot(x='Year', y='Temp', data=climate_data, ax=axs[2, 1])
axs[2, 1].set_title('Global Temperature Anomaly Over Time')

plt.tight_layout()
plt.show()

# Correlation between greenhouse gas concentrations and temperature anomalies
correlation_matrix = climate_data[['CO2', 'CH4', 'N2O', 'CFC-11', 'CFC-12', 'Temp']].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()

# Seasonal variation: CO2 and temperature anomalies
plt.figure(figsize=(14, 7))
sns.lineplot(x='Month', y='CO2', data=climate_data)
plt.title('Seasonal Variation of CO2 Concentration')
plt.show()

plt.figure(figsize=(14, 7))
sns.lineplot(x='Month', y='Temp', data=climate_data)
plt.title('Seasonal Variation of Temperature Anomalies')
plt.show()

# Impact of ENSO on temperature anomalies and greenhouse gas concentrations
plt.figure(figsize=(14, 7))
sns.lineplot(x='MEI', y='Temp', data=climate_data, label='Temp')
sns.lineplot(x='MEI', y='CO2', data=climate_data, label='CO2')
sns.lineplot(x='MEI', y='CH4', data=climate_data, label='CH4')
plt.title('Impact of ENSO (MEI) on Temperature Anomalies and Greenhouse Gas Concentrations')
plt.show()

# Effect of aerosols on global temperature anomalies
plt.figure(figsize=(14, 7))
sns.scatterplot(x='Aerosols', y='Temp', data=climate_data)
plt.title('Effect of Aerosols on Global Temperature Anomalies')
plt.show()
```

```python
# Line Chart: Trends Over Time
fig, axs = plt.subplots(3, 2, figsize=(15, 18))

# CO2 concentration over time
sns.lineplot(x='Year', y='CO2', data=climate_data, ax=axs[0, 0])
axs[0, 0].set_title('CO2 Concentration Over Time')

# CH4 concentration over time
sns.lineplot(x='Year', y='CH4', data=climate_data, ax=axs[0, 1])
axs[0, 1].set_title('CH4 Concentration Over Time')

# N2O concentration over time
sns.lineplot(x='Year', y='N2O', data=climate_data, ax=axs[1, 0])
axs[1, 0].set_title('N2O Concentration Over Time')

# CFC-11 concentration over time
sns.lineplot(x='Year', y='CFC-11', data=climate_data, ax=axs[1, 1])
axs[1, 1].set_title('CFC-11 Concentration Over Time')

# CFC-12 concentration over time
sns.lineplot(x='Year', y='CFC-12', data=climate_data, ax=axs[2, 0])
axs[2, 0].set_title('CFC-12 Concentration Over Time')

# Temperature anomaly over time
sns.lineplot(x='Year', y='Temp', data=climate_data, ax=axs[2, 1])
axs[2, 1].set_title('Global Temperature Anomaly Over Time')

plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap
correlation_matrix = climate_data[['CO2', 'CH4', 'N2O', 'CFC-11', 'CFC-12', 'Temp']].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
# Box Plot: Seasonal Variation in CO2 Concentration
plt.figure(figsize=(14, 7))
sns.boxplot(x='Month', y='CO2', data=climate_data)
plt.title('Seasonal Variation in CO2 Concentration')
plt.show()

# Box Plot: Seasonal Variation in Temperature Anomalies
plt.figure(figsize=(14, 7))
sns.boxplot(x='Month', y='Temp', data=climate_data)
plt.title('Seasonal Variation in Temperature Anomalies')
plt.show()

```

```python
# Scatter Plot: Impact of ENSO on Temperature Anomalies
plt.figure(figsize=(14, 7))
sns.scatterplot(x='MEI', y='Temp', data=climate_data)
plt.title('Impact of ENSO (MEI) on Temperature Anomalies')
plt.show()

# Scatter Plot: Impact of ENSO on CO2 Concentrations
plt.figure(figsize=(14, 7))
sns.scatterplot(x='MEI', y='CO2', data=climate_data)
plt.title('Impact of ENSO (MEI) on CO2 Concentrations')
plt.show()
```

```python
# Pair Plot: Multivariate Relationships
sns.pairplot(climate_data[['CO2', 'CH4', 'N2O', 'CFC-11', 'CFC-12', 'Temp']])
plt.suptitle('Pairwise Relationships in Climate Data', y=1.02)
plt.show()
```

```python
# Violin Plot: Distribution of Temperature Anomalies by Month
plt.figure(figsize=(14, 7))
sns.violinplot(x='Month', y='Temp', data=climate_data)
plt.title('Distribution of Temperature Anomalies by Month')
plt.show()
```

```python
# Calculate annual averages
annual_averages = climate_data.groupby('Year')[['CO2', 'CH4', 'N2O', 'CFC-11', 'CFC-12']].mean().reset_index()

# Bar Plot: Annual Average of CO2
plt.figure(figsize=(14, 7))
sns.barplot(x='Year', y='CO2', data=annual_averages)
plt.title('Annual Average of CO2 Concentrations')
plt.xticks(rotation=90)
plt.show()
```
